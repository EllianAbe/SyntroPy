**Short-Circuit Evaluation**
==============================

**Overview**
------------------------------
Short-circuit evaluation is a technique used in programming where the second operand of a logical operation is only evaluated if the first operand is not sufficient to determine the result.

**Function definition**

lets define a function that:
* receives a number X
* prints running X
* return True if X is zero, else False.

```python
def is_zero(x):
    print('running ', x)
    return x == 0
```

**OR Short Circuit**

In the case of the or operator, if the first operand is True, the second operand is not evaluated because the result is already determined to be True. A lot of useless processing can be ignored with the currect expression order. 

```python
print('only one of these will run (first is true)')
is_zero(0) or is_zero(1)

print()
print('all of these will run (first is false)')
is_zero(1) or is_zero(0)

print()
print('the evaluation will be stopped (second is true)')
is_zero(1) or is_zero(0) or is_zero(1)
```

**AND Short-circuit**

In the case of the and operator, if the first operand is False, the second operand is not evaluated because the result is already determined to be False.

```python
print('only one of these will run (first is false)')
is_zero(1) and is_zero(0)

print()
print('all of these will run (first is true)')
is_zero(0) and is_zero(2)

print()
print('the evaluation will be stopped (second is false)')
is_zero(0) and is_zero(1) and is_zero(2)
```

**Avoiding Short circuit**

Avoid short circuit if needed, you can simplily execute the functions before apply the logic operator.

```python
print('or')
first = is_zero(0)
second = is_zero(1)

first or second

print()

print('and')
first = is_zero(1)
second = is_zero(0)

first and second
```

Other way to do the same is build an iterable object with returns and call all / any function

all - all elements must be true (and)

any - any element must be true (or)

```python
all([is_zero(1), is_zero(0)])

any([is_zero(0), is_zero(1)])
```

#### Explanation

##### If the first Term is False, the second term will not be evaluated

##### using all to avoid short-circuit evaluation (AND)
